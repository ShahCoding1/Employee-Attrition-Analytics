```python
# ============================================================
# EMPLOYEE ATTRITION ANALYTICS
# Complete Google Colab Project
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency, ttest_ind

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully.")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# Upload your CSV file to Google Colab first

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully.")
print("File:", file_name)


# ============================================================
# 3. INITIAL DATA UNDERSTANDING
# ============================================================

print("\n========== DATASET SHAPE ==========")
print(df.shape)

print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== LAST 5 ROWS ==========")
display(df.tail())

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATA INFORMATION ==========")
df.info()


# ============================================================
# 4. MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

display(missing_table[missing_table["Missing Values"] > 0])

print("Total missing values:", df.isnull().sum().sum())


# ============================================================
# 5. DUPLICATE RECORDS
# ============================================================

print("\n========== DUPLICATES ==========")

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# ============================================================
# 6. REMOVE NON-INFORMATIVE COLUMNS
# ============================================================

# These columns contain no useful variation for analysis.

columns_to_remove = [
    "EmployeeCount",
    "EmployeeNumber",
    "Over18",
    "StandardHours"
]

df = df.drop(columns=columns_to_remove)

print("\nRemoved columns:")
print(columns_to_remove)

print("\nNew dataset shape:", df.shape)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")

display(df.describe().T)


# ============================================================
# 8. UNIQUE VALUES
# ============================================================

print("\n========== UNIQUE VALUES ==========")

for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique()[:20])


# ============================================================
# 9. TARGET VARIABLE — ATTRITION
# ============================================================

print("\n========== ATTRITION COUNT ==========")

attrition_counts = df["Attrition"].value_counts()

display(attrition_counts)


print("\n========== ATTRITION PERCENTAGE ==========")

attrition_percentage = (
    df["Attrition"]
    .value_counts(normalize=True)
    * 100
)

display(attrition_percentage.round(2))


# ============================================================
# 10. ATTRITION DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Attrition"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

total = len(df)

for container in ax.containers:
    labels = [
        f"{(bar.get_height()/total)*100:.1f}%"
        for bar in container
    ]
    ax.bar_label(container, labels=labels)

plt.tight_layout()
plt.show()


# ============================================================
# 11. ATTRITION BY DEPARTMENT
# ============================================================

department_rate = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION RATE BY DEPARTMENT ==========")
display(department_rate.round(2))


plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ============================================================
# 12. ATTRITION BY JOB ROLE
# ============================================================

jobrole_rate = (
    pd.crosstab(
        df["JobRole"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

jobrole_rate = jobrole_rate.sort_values(
    by="Yes",
    ascending=False
)

print("\n========== ATTRITION RATE BY JOB ROLE ==========")
display(jobrole_rate.round(2))


plt.figure(figsize=(11, 7))

sns.barplot(
    data=jobrole_rate.reset_index(),
    x="Yes",
    y="JobRole"
)

plt.title("Employee Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")

plt.tight_layout()
plt.show()


# ============================================================
# 13. OVERTIME ANALYSIS
# ============================================================

overtime_rate = (
    pd.crosstab(
        df["OverTime"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION RATE BY OVERTIME ==========")
display(overtime_rate.round(2))


plt.figure(figsize=(8, 5))

sns.barplot(
    data=overtime_rate.reset_index(),
    x="OverTime",
    y="Yes"
)

plt.title("Employee Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 14. CHI-SQUARE TEST — OVERTIME VS ATTRITION
# ============================================================

contingency_table = pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("\n========== CHI-SQUARE TEST ==========")
print("Chi-square statistic:", round(chi2, 4))
print("p-value:", p_value)
print("Degrees of freedom:", dof)

if p_value < 0.05:
    print(
        "Conclusion: There is evidence of an association "
        "between overtime and attrition."
    )
else:
    print(
        "Conclusion: There is insufficient evidence of an "
        "association between overtime and attrition."
    )


# ============================================================
# 15. AGE ANALYSIS
# ============================================================

print("\n========== AGE BY ATTRITION ==========")

age_summary = (
    df.groupby("Attrition")["Age"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(age_summary)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="Age"
)

plt.title("Age Distribution by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")

plt.tight_layout()
plt.show()


# ============================================================
# 16. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    kde=True,
    bins=20
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()


# ============================================================
# 17. MONTHLY INCOME ANALYSIS
# ============================================================

print("\n========== MONTHLY INCOME BY ATTRITION ==========")

income_summary = (
    df.groupby("Attrition")["MonthlyIncome"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(income_summary)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.show()


# ============================================================
# 18. OVERTIME + MONTHLY INCOME
# ============================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="OverTime",
    y="MonthlyIncome",
    hue="Attrition"
)

plt.title("Monthly Income by Overtime and Attrition")
plt.xlabel("Overtime")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.show()


# ============================================================
# 19. JOB SATISFACTION
# ============================================================

satisfaction_rate = (
    pd.crosstab(
        df["JobSatisfaction"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY JOB SATISFACTION ==========")
display(satisfaction_rate.round(2))


plt.figure(figsize=(9, 6))

sns.barplot(
    data=satisfaction_rate.reset_index(),
    x="JobSatisfaction",
    y="Yes"
)

plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 20. WORK-LIFE BALANCE
# ============================================================

worklife_rate = (
    pd.crosstab(
        df["WorkLifeBalance"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY WORK-LIFE BALANCE ==========")
display(worklife_rate.round(2))


plt.figure(figsize=(9, 6))

sns.barplot(
    data=worklife_rate.reset_index(),
    x="WorkLifeBalance",
    y="Yes"
)

plt.title("Attrition Rate by Work-Life Balance")
plt.xlabel("Work-Life Balance Level")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 21. YEARS AT COMPANY
# ============================================================

print("\n========== YEARS AT COMPANY ==========")

tenure_summary = (
    df.groupby("Attrition")["YearsAtCompany"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(tenure_summary)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="YearsAtCompany"
)

plt.title("Years at Company by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Years at Company")

plt.tight_layout()
plt.show()


# ============================================================
# 22. YEARS IN CURRENT ROLE
# ============================================================

role_tenure_rate = (
    df.groupby("Attrition")["YearsInCurrentRole"]
    .agg(["mean", "median", "std"])
    .round(2)
)

print("\n========== YEARS IN CURRENT ROLE ==========")
display(role_tenure_rate)


# ============================================================
# 23. YEARS SINCE LAST PROMOTION
# ============================================================

promotion_summary = (
    df.groupby("Attrition")["YearsSinceLastPromotion"]
    .agg(["mean", "median", "std"])
    .round(2)
)

print("\n========== YEARS SINCE LAST PROMOTION ==========")
display(promotion_summary)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="YearsSinceLastPromotion"
)

plt.title("Years Since Last Promotion by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Years Since Last Promotion")

plt.tight_layout()
plt.show()


# ============================================================
# 24. DISTANCE FROM HOME
# ============================================================

distance_summary = (
    df.groupby("Attrition")["DistanceFromHome"]
    .agg(["mean", "median", "std"])
    .round(2)
)

print("\n========== DISTANCE FROM HOME ==========")
display(distance_summary)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="DistanceFromHome"
)

plt.title("Distance From Home by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Distance From Home")

plt.tight_layout()
plt.show()


# ============================================================
# 25. BUSINESS TRAVEL
# ============================================================

travel_rate = (
    pd.crosstab(
        df["BusinessTravel"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY BUSINESS TRAVEL ==========")
display(travel_rate.round(2))


plt.figure(figsize=(9, 6))

sns.barplot(
    data=travel_rate.reset_index(),
    x="BusinessTravel",
    y="Yes"
)

plt.title("Attrition Rate by Business Travel")
plt.xlabel("Business Travel")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 26. MARITAL STATUS
# ============================================================

marital_rate = (
    pd.crosstab(
        df["MaritalStatus"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY MARITAL STATUS ==========")
display(marital_rate.round(2))


plt.figure(figsize=(9, 6))

sns.barplot(
    data=marital_rate.reset_index(),
    x="MaritalStatus",
    y="Yes"
)

plt.title("Attrition Rate by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 27. JOB LEVEL
# ============================================================

joblevel_rate = (
    pd.crosstab(
        df["JobLevel"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY JOB LEVEL ==========")
display(joblevel_rate.round(2))


plt.figure(figsize=(9, 6))

sns.barplot(
    data=joblevel_rate.reset_index(),
    x="JobLevel",
    y="Yes"
)

plt.title("Attrition Rate by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 28. GENDER ANALYSIS
# ============================================================

gender_rate = (
    pd.crosstab(
        df["Gender"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY GENDER ==========")
display(gender_rate.round(2))


# ============================================================
# 29. EDUCATION FIELD
# ============================================================

education_rate = (
    pd.crosstab(
        df["EducationField"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

education_rate = education_rate.sort_values(
    by="Yes",
    ascending=False
)

print("\n========== ATTRITION BY EDUCATION FIELD ==========")
display(education_rate.round(2))


plt.figure(figsize=(11, 6))

sns.barplot(
    data=education_rate.reset_index(),
    x="Yes",
    y="EducationField"
)

plt.title("Attrition Rate by Education Field")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Education Field")

plt.tight_layout()
plt.show()


# ============================================================
# 30. CORRELATION ANALYSIS
# ============================================================

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")

plt.tight_layout()
plt.show()


# ============================================================
# 31. CREATE BINARY ATTRITION VARIABLE
# ============================================================

df["Attrition_Binary"] = df["Attrition"].map({
    "Yes": 1,
    "No": 0
})

print("\nAttrition binary variable created.")


# ============================================================
# 32. CORRELATION WITH ATTRITION
# ============================================================

attrition_corr = (
    df.select_dtypes(include=np.number)
    .corr()["Attrition_Binary"]
    .drop("Attrition_Binary")
    .sort_values()
)

print("\n========== CORRELATION WITH ATTRITION ==========")

display(attrition_corr.round(3))


# ============================================================
# 33. TOP POSITIVE/NEGATIVE CORRELATIONS
# ============================================================

print("\n========== VARIABLES MOST ASSOCIATED WITH ATTRITION ==========")

display(
    attrition_corr
    .sort_values(
        key=lambda x: abs(x),
        ascending=False
    )
    .head(15)
    .round(3)
)


# ============================================================
# 34. T-TEST — AGE
# ============================================================

age_left = df.loc[
    df["Attrition"] == "Yes",
    "Age"
]

age_stayed = df.loc[
    df["Attrition"] == "No",
    "Age"
]

t_stat_age, p_age = ttest_ind(
    age_left,
    age_stayed,
    equal_var=False
)

print("\n========== T-TEST: AGE ==========")

print("t-statistic:", round(t_stat_age, 4))
print("p-value:", p_age)

if p_age < 0.05:
    print(
        "There is evidence of a difference in mean age "
        "between employees who left and stayed."
    )
else:
    print(
        "There is insufficient evidence of a difference "
        "in mean age between the two groups."
    )


# ============================================================
# 35. T-TEST — MONTHLY INCOME
# ============================================================

income_left = df.loc[
    df["Attrition"] == "Yes",
    "MonthlyIncome"
]

income_stayed = df.loc[
    df["Attrition"] == "No",
    "MonthlyIncome"
]

t_stat_income, p_income = ttest_ind(
    income_left,
    income_stayed,
    equal_var=False
)

print("\n========== T-TEST: MONTHLY INCOME ==========")

print("t-statistic:", round(t_stat_income, 4))
print("p-value:", p_income)

if p_income < 0.05:
    print(
        "There is evidence of a difference in mean monthly "
        "income between employees who left and stayed."
    )
else:
    print(
        "There is insufficient evidence of a difference "
        "in mean monthly income between the two groups."
    )


# ============================================================
# 36. CHI-SQUARE TEST FUNCTION
# ============================================================

def chi_square_test(data, variable, target="Attrition"):

    table = pd.crosstab(
        data[variable],
        data[target]
    )

    chi2, p, dof, expected = chi2_contingency(table)

    return {
        "Variable": variable,
        "Chi-Square": chi2,
        "p-value": p,
        "Significant": "Yes" if p < 0.05 else "No"
    }


# ============================================================
# 37. MULTIPLE CHI-SQUARE TESTS
# ============================================================

categorical_variables = [
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime"
]

chi_results = []

for variable in categorical_variables:

    result = chi_square_test(
        df,
        variable
    )

    chi_results.append(result)

chi_results_df = pd.DataFrame(
    chi_results
)

chi_results_df["Chi-Square"] = (
    chi_results_df["Chi-Square"].round(4)
)

chi_results_df["p-value"] = (
    chi_results_df["p-value"].round(6)
)

print("\n========== CHI-SQUARE RESULTS ==========")

display(
    chi_results_df.sort_values(
        "p-value"
    )
)


# ============================================================
# 38. ATTRITION BY AGE GROUP
# ============================================================

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56+"
    ]
)

agegroup_rate = (
    pd.crosstab(
        df["AgeGroup"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY AGE GROUP ==========")

display(agegroup_rate.round(2))


plt.figure(figsize=(10, 6))

sns.barplot(
    data=agegroup_rate.reset_index(),
    x="AgeGroup",
    y="Yes"
)

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# 39. ATTRITION BY TENURE GROUP
# ============================================================

df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, 20, 100],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)

tenure_rate = (
    pd.crosstab(
        df["TenureGroup"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY TENURE GROUP ==========")

display(tenure_rate.round(2))


plt.figure(figsize=(10, 6))

sns.barplot(
    data=tenure_rate.reset_index(),
    x="TenureGroup",
    y="Yes"
)

plt.title("Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ============================================================
# 40. ATTRITION BY INCOME GROUP
# ============================================================

df["IncomeGroup"] = pd.qcut(
    df["MonthlyIncome"],
    q=4,
    labels=[
        "Low Income",
        "Lower-Middle Income",
        "Upper-Middle Income",
        "High Income"
    ]
)

income_rate = (
    pd.crosstab(
        df["IncomeGroup"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

print("\n========== ATTRITION BY INCOME GROUP ==========")

display(income_rate.round(2))


plt.figure(figsize=(10, 6))

sns.barplot(
    data=income_rate.reset_index(),
    x="IncomeGroup",
    y="Yes"
)

plt.title("Attrition Rate by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ============================================================
# 41. TOP ATTRITION PATTERNS
# ============================================================

print("\n")
print("=" * 60)
print("KEY ATTRITION PATTERNS")
print("=" * 60)

overall_attrition = (
    df["Attrition"]
    .value_counts(normalize=True)
    .get("Yes", 0) * 100
)

print(
    f"\nOverall attrition rate: "
    f"{overall_attrition:.2f}%"
)


if "Yes" in overtime_rate.columns:

    overtime_max = overtime_rate["Yes"].idxmax()

    overtime_max_rate = overtime_rate.loc[
        overtime_max,
        "Yes"
    ]

    print(
        f"Highest overtime-related attrition group: "
        f"{overtime_max} ({overtime_max_rate:.2f}%)"
    )


jobrole_max = jobrole_rate["Yes"].idxmax()

print(
    f"Job role with highest observed attrition rate: "
    f"{jobrole_max}"
)

print(
    f"Observed attrition rate for this role: "
    f"{jobrole_rate.loc[jobrole_max, 'Yes']:.2f}%"
)


department_max = department_rate["Yes"].idxmax()

print(
    f"Department with highest observed attrition rate: "
    f"{department_max}"
)

print(
    f"Observed attrition rate for this department: "
    f"{department_rate.loc[department_max, 'Yes']:.2f}%"
)


# ============================================================
# 42. SUMMARY TABLE
# ============================================================

summary = pd.DataFrame({

    "Metric": [
        "Total Employees",
        "Employees Who Left",
        "Employees Who Stayed",
        "Overall Attrition Rate",
        "Average Age",
        "Average Monthly Income",
        "Average Years at Company"
    ],

    "Value": [
        len(df),
        (df["Attrition"] == "Yes").sum(),
        (df["Attrition"] == "No").sum(),
        round(overall_attrition, 2),
        round(df["Age"].mean(), 2),
        round(df["MonthlyIncome"].mean(), 2),
        round(df["YearsAtCompany"].mean(), 2)
    ]
})

print("\n========== PROJECT SUMMARY ==========")

display(summary)


# ============================================================
# 43. SAVE CLEANED DATASET
# ============================================================

# Remove helper columns if desired before exporting

export_df = df.copy()

export_df.to_csv(
    "employee_attrition_cleaned.csv",
    index=False
)

print("\nCleaned dataset saved as:")
print("employee_attrition_cleaned.csv")


# ============================================================
# 44. DOWNLOAD CLEANED DATASET
# ============================================================

files.download(
    "employee_attrition_cleaned.csv"
)


# ============================================================
# 45. FINAL MESSAGE
# ============================================================

print("\n")
print("=" * 60)
print("EMPLOYEE ATTRITION ANALYTICS COMPLETED")
print("=" * 60)

print("""
The project included:

✓ Data loading
✓ Data cleaning
✓ Missing-value analysis
✓ Duplicate analysis
✓ Descriptive statistics
✓ Attrition analysis
✓ Department analysis
✓ Job-role analysis
✓ Overtime analysis
✓ Age analysis
✓ Income analysis
✓ Job satisfaction analysis
✓ Work-life balance analysis
✓ Employee tenure analysis
✓ Promotion analysis
✓ Distance-from-home analysis
✓ Business travel analysis
✓ Correlation analysis
✓ Chi-square statistical tests
✓ T-tests
✓ Age-group analysis
✓ Tenure-group analysis
✓ Income-group analysis
✓ Summary statistics
✓ Cleaned dataset export
""")
```
